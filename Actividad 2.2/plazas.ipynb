```python
#Importamos las librerias pandas, numpy y matplotlib respectivamente
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
#Cargar el archivo Excel
data = pd.read_excel("Plazas.xlsx")

data.head()
```

```python
#Ver número de filas y columnas
data.shape
```

```python
#Crear una copia del dataframe original
data1 = data.copy()
```

```python
#Conservar únicamente registros que tengan un nombre
data1 = data1[data1["Nombre"].notnull()]
```

```python
#Reiniciar índices
data1 = data1.reset_index(drop=True)
```

```python
#Comprobación

data1.shape
```

```python
#Identificar valores nulos por columna

valores_nulos = data1.isnull().sum()
valores_nulos
```

```python
#Eliminar columnas que se encuentran completamente vacías

data1 = data1.drop(["Telefono", "Unnamed: 17", "Unnamed: 18"], axis=1)
```

```python
#Verificar valores nulos por columna

data1.isnull().sum()
```

```python
#Crear otra copia para comenzar la sustitución de valores nulos

data2 = data1.copy()
```

```python
#Convertir la columna a tipo fecha
data2["Fecha Origen"] = pd.to_datetime(data2["Fecha Origen"], errors="coerce")

#Sustituir valores nulos por valores no nulos hacia adelante
#Forward fill

#Se utiliza este método porque los registros están ordenados por fecha
#y solo existen 2 valores nulos, por lo que se toma la fecha anterior registrada.

data2["Fecha Origen"] = data2["Fecha Origen"].ffill()
```

```python
#Sustituir valores nulos por False

#Se utiliza False porque esta columna indica si hubo seguimiento o no.
#Un valor nulo se interpreta como que no se registró seguimiento.

data2["Seguimiento"] = data2["Seguimiento"].fillna(False)
```

```python
#Sustituir valores nulos por False

#Se utiliza False porque esta columna indica si se realizó una cita efectiva.
#Si no existe un valor registrado, se considera que no se concretó la cita.

data2["Cita efetiva "] = data2["Cita efetiva "].fillna(False)
```

```python
#Sustituir valores nulos por False

#Se utiliza False porque PDM es una variable de tipo verdadero o falso.
#Los valores nulos se consideran como una actividad que no fue registrada como realizada.

data2["PDM"] = data2["PDM"].fillna(False)
```

```python
#Sustituir valores nulos por False

#Se utiliza False porque SDC registra si se realizó o no esta actividad.
#Cuando no existe información se considera que no fue registrada como realizada.

data2["SDC"] = data2["SDC"].fillna(False)
```

```python
#Sustituir valores nulos por un string

#Se utiliza "Sin información" porque no existe un estatus registrado
#y no es correcto asumir algún estado específico.

data2["Estatus"] = data2["Estatus"].fillna("Sin información")
```

```python
#Sustituir valores nulos por "No aplica"

#Se utiliza este valor porque si hubo seguimiento,
#no existe una razón que justifique su ausencia.

data2["¿Por que no hay seguimiento?"] = data2["¿Por que no hay seguimiento?"].fillna("No aplica")
```

```python
#Sustituir valores nulos por "No aplica"

#Se utiliza este valor porque una celda vacía puede indicar que
#no fue necesario registrar una razón para no visitar la agencia.

data2["¿Por que no ha visitado la agencia?"] = data2["¿Por que no ha visitado la agencia?"].fillna("No aplica")
```

```python
#Sustituir valores nulos por "No aplica"

#Se utiliza este valor porque no todos los registros requieren
#una justificación relacionada con la prueba de manejo.

data2["¿Por que no hubo prueba de manejo ?"] = data2["¿Por que no hubo prueba de manejo ?"].fillna("No aplica")
```

```python
#Sustituir valores nulos por "No aplica"

#Se utiliza este valor porque una solicitud de crédito
#no necesariamente corresponde a todos los clientes.

data2["¿Por que no hubo solicitud de credito?"] = data2["¿Por que no hubo solicitud de credito?"].fillna("No aplica")
```

```python
#Sustituir valores nulos por "No aplica"

#Se utiliza este valor porque no todos los registros se encuentran
#en una etapa donde sea necesario justificar la falta de formalización.

data2["¿Por que no se ha formalizado la compra?"] = data2["¿Por que no se ha formalizado la compra?"].fillna("No aplica")
```

```python
#Sustituir valores nulos por "Sin información"

#Se utiliza este valor porque la columna contiene observaciones adicionales
#de los clientes y no es correcto inventar una observación cuando no existe información.

data2["Unnamed: 16"] = data2["Unnamed: 16"].fillna("Sin información")
```

```python
#Eliminar la columna Unnamed: 19

#Se elimina porque 302 de sus 303 registros son valores nulos
#y la columna no cuenta con un encabezado que permita identificar su propósito.

data2 = data2.drop(["Unnamed: 19"], axis=1)
```

```python
#Verificamos valores nulos por columna
data2.isnull().sum()
```

```python
data2.to_csv("Plazas_Limpio.csv", index=False)
```
