```python
#Importamos las librerias pandas, numpy y matplotlib respectivamente
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
#Carga desde un archivo .csv sin indice
df = pd.read_csv('Citas_Digital_Filtrado.csv')
df.info()
```

```python
#Eliminar columnas innecesarias
#axis = 1 para columnas
#axis = 0 para filas
df1= df.drop(['Unnamed: 15'], axis=1)
df1
```

```python
#Identificar valores nulos por columna
valores_nulos=df1.isnull().sum()
valores_nulos
```

```python
#Identificar valores nulos por dataframe
valores_nulos=df1.isnull().sum().sum()
valores_nulos
```

```python
#Realizamos una copia del dataframe
df2=df1.copy()
```

```python
#Quinto método de sustitución de valores nulos
#Sustituir valores nulos por valores no nulos hacia adelante  "forward fill" ("ffill")
#Filtro por columnas
df2["Asesor Asignado"] =df2["Asesor Asignado"].ffill()
```

```python
#Corroboramos valores nulos
valores_nulos=df2.isnull().sum()
valores_nulos
```

```python
#Sexto método de sustitución de valores nulos
#Sustituir valores nulos por valores no nulos hacia atrás backward fill" ("bfill")
#Filtro por columnas
df2["Nombre Cliente "] =df2["Nombre Cliente "].bfill() 
```

```python
#Corroboramos valores nulos
valores_nulos=df2.isnull().sum()
valores_nulos
```

```python
#Sustituir valores nulos por valores no nulos hacia adelante forward fill" ("ffill")
#Filtro por columnas
df2["BDC "] =df2["BDC "].ffill()
```

```python
#Corroboramos valores nulos
valores_nulos=df2.isnull().sum()
valores_nulos
```

```python
#Sustituir valores nulos por valores no nulos hacia atrás backward fill" ("bfill")
df2["Estatus de Lead"] =df2["Estatus de Lead"].bfill() 
```

```python
#Corroboramos valores nulos
valores_nulos=df2.isnull().sum()
valores_nulos
```

```python
#Sustituir valores nulos por valores no nulos hacia adelante forward fill" ("ffill")
#Filtro por columnas
df2["¿Por que no ha visitado la agencia?"] =df2["¿Por que no ha visitado la agencia?"].ffill()    
```

```python
#Corroboramos valores nulos
valores_nulos=df2.isnull().sum()
valores_nulos
```

```python
#Corroboramos valores nulos
valores_nulos=df2.isnull().sum()
valores_nulos
```

```python
df2["Venta"]= df2["Venta"].fillna(0) 
df2
```

```python
df2["SDC"]= df2["SDC"].fillna(0) 
```

```python
df2["PDM"]= df2["PDM"].fillna(0) 
```

```python
#Corroboramos valores nulos
valores_nulos=df2.isnull().sum()
valores_nulos
```

```python
#Sustituir valores nulos por valores no nulos hacia atrás backward fill" ("bfill")
df2["Fecha"] =df2["Fecha"].bfill() 
```

```python
#Cuarto método de sustitución de valores nulos
#Sustituir valores nulos por un string en  concreto 
df2["Telefono "] =df2["Telefono "].fillna("Sin teléfono") 
```

```python
#Convertir la columna a número (float)
df2["Potencial de compra"] = pd.to_numeric(df2["Potencial de compra"], errors='coerce')

#Segundo método de sustitución de valores nulos
#Sustituir valores nulos con mediana
df2["Potencial de compra"]=df2["Potencial de compra"].fillna(round(df2["Potencial de compra"].median(),1))
```

```python
#Corroboramos valores nulos
valores_nulos=df2.isnull().sum()
valores_nulos
```
